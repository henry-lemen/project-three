# Chicago Traffic Data

This is the notebook used to clean up the data used for this project. The original file was very large so it had to be seperated by year in order to be uploaded to github.

```python
# imports
import pandas as pd
from pathlib import Path
import datetime
```

# Original File

```python
# Assign file to variable
file = Path("../data/traffic_crashes.csv")
```

```python
# Create dataframe and view head
crash_data_df = pd.read_csv(file)
crash_data_df.head()
```

```python
# Check dataframe information
crash_data_df.info()
```

```python
# Reduce columns
cleaned_crash_data = crash_data_df[['RD_NO', 'CRASH_DATE', 'POSTED_SPEED_LIMIT', 'WEATHER_CONDITION', 'LIGHTING_CONDITION', 'FIRST_CRASH_TYPE', 'TRAFFICWAY_TYPE', 
                                    'ROADWAY_SURFACE_COND', 'ROAD_DEFECT', 'CRASH_TYPE', 'DAMAGE', 'PRIM_CONTRIBUTORY_CAUSE', 'SEC_CONTRIBUTORY_CAUSE', 'STREET_NAME', 
                                    'CRASH_HOUR', 'CRASH_DAY_OF_WEEK', 'CRASH_MONTH', 'LATITUDE', 'LONGITUDE', 'LOCATION']]
cleaned_crash_data.head()
```

```python
# Drop incomplete columns and check that columns have same amount of rows
cleaned_df = cleaned_crash_data.dropna()
cleaned_df.info()
```

```python
# Convert date to datetime
cleaned_df['CRASH_DATE'] = pd.to_datetime(cleaned_df['CRASH_DATE'])
cleaned_df.info()
```

```python
# Order the accidents by date and select desired years
selected_years = [2018, 2019, 2020, 2021, 2022]
sorted_df = cleaned_df[cleaned_df['CRASH_DATE'].dt.year.isin(selected_years)]
sorted_df = sorted_df.sort_values(by='CRASH_DATE', ascending=True)
sorted_df.head()
```

```python
# Select desired columns
final_df = sorted_df[['CRASH_DATE', 'POSTED_SPEED_LIMIT', 'WEATHER_CONDITION', 'LIGHTING_CONDITION', 'FIRST_CRASH_TYPE', 
                                    'ROADWAY_SURFACE_COND', 'ROAD_DEFECT', 'CRASH_TYPE', 'DAMAGE', 'PRIM_CONTRIBUTORY_CAUSE', 'SEC_CONTRIBUTORY_CAUSE', 'STREET_NAME', 
                                    'CRASH_HOUR', 'CRASH_DAY_OF_WEEK', 'CRASH_MONTH', 'LATITUDE', 'LONGITUDE']]
final_df = final_df.reset_index(drop=True)
final_df.head()
```

### Seperate data by year and write to CSV

```python
df_2018 = final_df[final_df['CRASH_DATE'].dt.year == 2018]
df_2019 = final_df[final_df['CRASH_DATE'].dt.year == 2019]
df_2020 = final_df[final_df['CRASH_DATE'].dt.year == 2020]
df_2021 = final_df[final_df['CRASH_DATE'].dt.year == 2021]
df_2022 = final_df[final_df['CRASH_DATE'].dt.year == 2022]
```

```python
df_2018.to_csv('../yearly_crash_data/df_2018.csv', index=False, header=True)
df_2019.to_csv('../yearly_crash_data/df_2019.csv', index=False, header=True)
df_2020.to_csv('../yearly_crash_data/df_2020.csv', index=False, header=True)
df_2021.to_csv('../yearly_crash_data/df_2021.csv', index=False, header=True)
df_2022.to_csv('../yearly_crash_data/df_2022.csv', index=False, header=True)

```
